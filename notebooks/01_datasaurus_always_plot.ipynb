### Imports

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.size'] = 16
plt.rcParams['figure.figsize'] = (8, 5)
```

## The story

Several different datasets can share nearly identical means, standard deviations, and correlations — yet look completely different when you plot them. **Always plot.**

## 1. Load the evidence with `pd.read_csv()`

The file is the witness. Start by opening it.

```python
df = pd.read_csv("../data/always_plot_demo.csv")
```

## 2. First glance with `df.head()`

What does one row look like?

```python
df.head()
```

## 3. Last glance with `df.tail()`

Catch weird endings, appended notes, or format changes.

```python
df.tail()
```

## 4. Random glance with `df.sample()`

If rows are sorted, `head()` can stage-manage the evidence. `sample()` breaks the staging.

```python
df.sample(5, random_state=42)
```

## 5. How much evidence? `df.shape`

```python
df.shape
```

## 6. Name the variables with `df.columns`

```python
list(df.columns)
```

## 7. Schema check with `df.info()`

`info()` combines non-null counts with types — fastest first audit.

```python
df.info()
```

## 8. Data types with `df.dtypes`

```python
df.dtypes
```

## 9. Isolate numeric columns with `df.select_dtypes()`

```python
numeric = df.select_dtypes(include="number")
numeric.head()
```

## 10. Look at summary statistics for each shape

Notice how similar the summaries are — yet the data is wildly different (we will see that next).

```python
df.groupby("dataset")[["x", "y"]].agg(["mean", "std"]).round(2)
```

## 11. Correlations per dataset

```python
df.groupby("dataset")[["x", "y"]].corr().round(2)
```

## 12. Now plot each dataset

The story lands when you see the shapes.

```python
for name, part in df.groupby("dataset"):
    ax = part.plot(kind="scatter", x="x", y="y", title=name)
    plt.show()
```

## Discussion

- What would a report of means and standard deviations hide here?
- Which dataset would you flag as suspicious if you only had the summary table?


## Mini-lab: the plotting habit

Build a fast EDA opening ritual:

```python
df = pd.read_csv("../data/always_plot_demo.csv")
print(df.shape)
print(df.dtypes)
print(df.sample(5, random_state=42))
df.query("dataset == 'line'").plot(kind="scatter", x="x", y="y")
```
